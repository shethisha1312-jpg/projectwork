```python
import numpy as np

class DataAnalytics:
    
    def __init__(self,array=None):
        if array is None:
            self.array = np.array([])
            
        else:
            self.array = np.array(array)
            
    def __set_array(self,array):
        self.array = np.array(array)
        
    def __display_array(self,arr):
        print("\n Array:")
        print(arr)
        
    @classmethod
    def create_from_list(cls,data):
        return cls(np.array(data))
    
    @staticmethod
    def show_title():
        print("Numpy ANalyser") 
    
    def display(self):
        print("Current Array:")
        print(self.array)
        
    def create_array (self):
        
        print("Select the type of array to create:")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")
        
        choice = int(input("Enter your choice:"))
        
        if choice == 1:
            n = int(input("Enter the number of elements: "))

            values = list(map(int, input(f"Enter {n} elements separated by space: ").split()))
            self.__set_array(values)

            print("Array created successfully:")
            self.__display_array(self.array)
            
        elif choice == 2:
            rows = int(input("Enter the numbers of rows:"))
            cols = int(input("Enter the numbers of columns:"))
            
            values = list(map(int,input(f"Enter {rows * cols} elements separated by space:").split()))
            
            self.__set_array(np.array(values).reshape(rows,cols))
            
            
            print("Array created successfully:")
            self.__display_array(self.array)
            
        elif choice == 3:
            
            layers = int(input("enter number of layers:"))
            rows = int(input("Enter number of rows:"))
            cols = int(input("Enter number of columns:"))
            
            elements = list(map(int,input(f"Enter {layers * rows * cols } elemnts seperated by space:").split()))
            
            self.array = np.array(elements).reshape(layers , rows , cols)
            
            print("Array created successfully:")
            self.__display_array(self.array)
        
            
            
        else:
            
            print("Invalid Choice")
            
    def indexing_slicing(self):
        
        if self.array is None:
            
            print("Please create an array firsr.")
            return 
        
        
        print("Choose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")
        
        choice = int(input("Enter your choice :"))
        
        if choice == 1:
            
            if self.array.ndim == 1:
                
                index = int(input("Enter index:"))
                print("Elements:",self.array[index])
                
            elif self.array.ndim == 2:
                
                
                row = int(input("Enter rows index:"))
                
                col = int(input("Enter columns index:"))
                print("Element:", self.array[row, col])
                
                
            elif self.array.ndim == 3:
                
                
                layer = int(input("Enter layer index: "))
                row = int(input("Enter row index: "))
                col = int(input("Enter column index: "))

                print("Element:", self.array[layer, row, col])
                
                
        elif choice == 2:
            
            if self.array.ndim == 1:
                start = int(input("Enter start index: "))
                end = int(input("Enter end index: "))

                print("Sliced Array:")
                print(self.array[start:end])
                
            elif self.array.ndim == 2:
                
                start_row = int(input("Enter row start: "))

                end_row = int(input("Enter row end: "))

                start_col = int(input("Enter column start: "))

                end_col = int(input("Enter column end: "))

                result = self.array[start_row:end_row,
                                    start_col:end_col
                                    ]

                print("\nSliced Array:")
                print(result)
                
            
            elif self.array.ndim == 3:
                start_layer = int(input("Enter layer start: "))
                end_layer = int(input("Enter layer end: "))
                start_row = int(input("Enter row start: "))
                end_row = int(input("Enter row end: "))
                start_col = int(input("Enter column start: "))
                end_col = int(input("Enter column end: "))

                result = self.array[
                    start_layer:end_layer,
                    start_row:end_row,
                    start_col:end_col
                ]

                print("\nSliced Array:")
                print(result)
                
            elif choice == 3:
                return 
            
            else:
                print("Invalid choice!")
                
    def mathematical_operation(self):
        
        if self.array is None:
            
            print("Please create an array first.")
            return
        
        print("Choose a mathematical operation:")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication ")
        
        choice = int(input("Enter your choice:"))
        
        if choice in [1,2,3,4]:
            
            size = self.array.size
            elements = list(map(int,input(f"\nEnter the same-size array elements "
                                          f"({size} elements separated by space): ").split()))
            
            
            
            second_array = np.array(elements).reshape(self.array.shape)
            
            print("\nOriginal Array:")
            print(self.array)

            print("\nSecond Array:")
            print(second_array)
            
            if choice == 1:
                result = self.array + second_array
                print(" Result of Addition:")
                print(result)
                
            elif choice == 2:
    
                result = self.array - second_array

                print("Result of Subtraction:")
                print(result)
                
            elif choice == 3:
    
                result = self.array * second_array

                print("Result of Multiplication:")
                print(result)
            
            elif choice == 4:
                
                result = self.array / second_array

                print("\nResult of Division:")
                print(result)
                
        elif choice == 5:
            if self.array.ndim != 1:
                print("Dot Product is demonstrated using 1D array.")
                return 
            
            elements = list(map(int,input(f"Enter {self.array.size} elements:").split()))
            
            second_array = np.array(elements)
            result = np.dot(self.array,
                            second_array
                            )
            print("Dot Product:")
            print(result)
            
        elif choice == 6:
            if self.array.ndim != 2:
                print("Matrix multiplication requiers 2D array.")
                return
            
            rows = self.array.shape[1]
            
            elements = list(map(int,input(f"Enter emenets for second matrix"
                                          f"(number of rows must be {rows}):").split()))
            
            cols = int(input("Enter number of columns: "))

            second_array = np.array(elements).reshape(rows,cols)
            
            result = np.matmul(self.array,second_array)
            

            print("\nFirst Matrix:")
            print(self.array)

            print("\nSecond Matrix:")
            print(second_array)

            print("\nMatrix Multiplication:")
            print(result)
            
        else:
            
            print("Invalid Choice !")
    def combine_split(self):
        
        if self.array is None:
            print("Please create an array first.")
            return
        
        print("Choose an option:")
        print("1. Combine Arrays")
        print("2. Split Array")
        
        choice = int(input("Enter your choice:"))
        
        if choice == 1:
            
            elements = list(map(int,input(f"Enter the elements of another to combine"
                                        f"({self.array.size} elements seprated by space):").split()))
            
            second_array = np.array(elements).reshape(self.array.shape)
            
            print("\nOrigina Array:")
            print(self.array)
            
            result = np.vstack((self.array, second_array))

            print("\nCombined Array (Vertical Stack):")
            print(result)

            print("\nArrays combined successfully!")
            
        elif choice == 2:
            
            parts = int(input("\n Enter number of parts to split:"))
            
            try:
                result = np.array_split(self.array,parts)
                print("\n split Arrays:")
                
                for i in range(len(result)):
                    print("\n Part ", i + 1 )
                    print(result[i])
                    
                print("\n Array split successfully")
                
            except Exception as e:
                print("Invalid choice.")
                
    def search_sort_filter(self):
            
            if self.array is None:
                print("Please create an array first.")
                return
            
            print("\n Choose an option")
            print("1. Search a value")
            print("2. Sort the array")
            print("3. Filter values")
            
            choice = int(input("Enter your choice :"))
            
            if choice == 1:
                
                value = int(input("Enter value to search: "))

                positions = np.where(self.array == value)
            

                if positions[0].size > 0:

                    print("\nValue found at position(s):")
                    print(positions)

                else:

                    print("\nValue not found.")
                
            elif choice == 2:
                print("\nOriginal Array:")
                print(self.array)

                print("\nChoose sorting order:")
                print("1. Ascending")
                print("2. Descending")

                order = int(input("Enter your choice: "))
                
                if order == 1:
    
                    if self.array.ndim == 2:
                        result = np.sort(self.array,axis=1)
                    else:
                        result = np.sort(self.array)

                elif order == 2:

                    if self.array.ndim == 2:
                        result = np.sort(self.array,axis=1)[:, ::-1]
                    else:
                        result = np.sort(self.array)[::-1]

                else:

                    print("Invalid choice!")
                    return

                print("\nSorted Array:")
                print(result)

                print("\nArray sorted successfully!")
                
            elif choice == 3:
        
                value = int(input("\nEnter value. Show elements greater than this value: "))

                result = self.array[
                    self.array > value
                ]

                print("\nFiltered Values:")
                print(result)

    
            else:
                print("Invalid choice!")
                
    def aggregates_statistics(self):
    
            if self.array is None:
                print("Please create an array first.")
                return

            print("\nChoose an aggregate/statistical operation:")
            print("1. Sum")
            print("2. Mean")
            print("3. Median")
            print("4. Standard Deviation")
            print("5. Variance")
            print("6. Minimum")
            print("7. Maximum")
            print("8. Percentile")
            print("9. Correlation Coefficient")

            choice = int(input("Enter your choice: "))
            
            if choice == 1:
    
                print("\n Array:")
                print(self.array)

                print("\nSum of Array:",np.sum(self.array))
                
            elif choice == 2:
        
                print("\n Array:")
                print(self.array)

                print("\nMean of Array:",np.mean(self.array))
                
            elif choice == 3:
    
                print("\n Array:")
                print(self.array)

                print("\nMedian of Array:",np.median(self.array))

            elif choice == 4:
    
                print("\n Array:")
                print(self.array)

                print("\nStandard Deviation:",np.std(self.array))
            
            elif choice == 5:
    
                print("\n Array:")
                print(self.array)

                print("\nVariance:",np.var(self.array))

            elif choice == 6:

                print("\nMinimum Value:",np.min(self.array))
                
            elif choice == 7:
        
                print("\nMaximum Value:",np.max(self.array))  
            
            elif choice == 8:
    
                p = float(input("\nEnter percentile (example 50): "))

                print("\nPercentile:",np.percentile(self.array,p)) 
            elif choice == 9:
    
                if self.array.ndim != 1:

                    print("Correlation is demonstrated using 1D arrays.")
                    return

                elements = list(map(int, input(f"Enter {self.array.size} elements: ").split()))

                second_array = np.array(elements)

                result = np.corrcoef(self.array,second_array)
                

                print("\nCorrelation Coefficient:")
                print(result)

            else:
              print("Invalid choice!")      
              
    @classmethod
    def project_name(cls):
        print("\nNumPy Analyzer")
                    
    @staticmethod
    def welcome():
        print("\nWelcome to the NumPy Analyzer!")  
            

DataAnalytics.welcome()
DataAnalytics.project_name()

obj = DataAnalytics()

while True:
           
    print("\n====================================")
    print("Choose an option:")
    print("1. Create a NumPy Array")
    print("2. Perform Mathematical Operations")
    print("3. Combine or Split Arrays")
    print("4. Search, Sort, or Filter Arrays")
    print("5. Compute Aggregates and Statistics")
    print("6. Indexing and Slicing")
    print("7. Exit")
    print("====================================")
    
    choice = int(input("Enter your choice :"))
    
    if choice == 1:
        obj.create_array()
        
    elif choice == 2:
        obj.mathematical_operation()
    elif choice  == 3:
        obj.combine_split()
    elif choice == 4:
        obj.search_sort_filter()
    elif choice == 5:
        obj.aggregates_statistics()
    elif choice == 6:
        obj.indexing_slicing()
    elif choice == 7:
        print("\n Thank you for using the Numpy Analyzer! Goodbye! ")
        break
    else:
        print("Invalid choice! Please try agin.")

                                
                                
                        
                        
                    
            
            
        
            
            
            
            
                
            
                

                
                
                
                
        

                  
```
